# Aula 02 — Agrupamento de dados

Mesma base da Aula 01 (planilha SEEG, aba `GEE Estados`). O arquivo mora em `../01_conhecendo_os_dados/dados/` — não duplicamos os 78 MB aqui.

**Padrão bruto → tratado:** a leitura grava em `emissoes_brutas` (roda uma vez, é a célula lenta) e o tratamento deriva `emissoes_gases` a partir dela. Como o tratamento nunca sobrescreve sua própria fonte, ele pode ser reexecutado à vontade sem `KeyError`.

```python
import pandas as pd
```

```python
# Célula LENTA — roda uma vez só. O bruto nunca é alterado depois daqui.
# A planilha fica na pasta da Aula 01 — caminho relativo evita duplicar 78 MB
emissoes_brutas = pd.read_excel(
    '../01_conhecendo_os_dados/dados/1-SEEG10_GERAL-BR_UF_2022.10.27-FINAL-SITE.xlsx',
    sheet_name='GEE Estados'
)
emissoes_brutas.shape
```

```python
# Tratamento da Aula 01: só linhas de Emissão (Remoção é negativa e Bunker não tem Estado)
# Lê de emissoes_brutas e escreve em emissoes_gases → reexecutável quantas vezes quiser
# .copy() evita o SettingWithCopyWarning nas colunas que criaremos nesta aula
emissoes_gases = emissoes_brutas[emissoes_brutas['Emissão / Remoção / Bunker'] == 'Emissão'].copy()
emissoes_gases = emissoes_gases.drop(columns='Emissão / Remoção / Bunker')

# Validação: (94748, 62) — 94.748 linhas de emissão, 62 colunas após o drop
emissoes_gases.shape
```

```python
emissoes_gases.loc[:,'Nível 1 - Setor':'Produto'].columns
```

```python
colunas_info = list(emissoes_gases.loc[:,'Nível 1 - Setor':'Produto'].columns)
colunas_info
```

```python
emissoes_gases.loc[:,1970:2021].columns
```

```python
colunas_emissao = list(emissoes_gases.loc[:,1970:2021].columns)
colunas_emissao
```

```python
emissoes_gases.melt(id_vars = colunas_info, value_vars = colunas_emissao, var_name = 'Ano', value_name = 'Emissão')

```

```python
emissoes_por_ano = emissoes_gases.melt(id_vars = colunas_info, value_vars = colunas_emissao, var_name = 'Ano' , value_name = 'Emissão')
```

```python
emissoes_por_ano.groupby('Gás')
```

```python
emissoes_por_ano.groupby('Gás').groups
```

```python
emissoes_por_ano.groupby('Gás').get_group('CO2 (t)')
```

```python
emissoes_por_ano.groupby('Gás')[['Emissão']].sum()
```

```python
emissao_por_gas = emissoes_por_ano.groupby('Gás')[['Emissão']].sum().sort_values(by='Emissão', ascending=False)
emissao_por_gas
```

```python
emissao_por_gas.plot(kind='barh', figsize =(10,6));
```

```python
emissao_por_gas.iloc[0:9]
```

```python
print(f'A emissão de CO2 corresponde a {float((emissao_por_gas.iloc[0:9].sum() / emissao_por_gas.sum()) * 100):.2f} % de emissão total de gases estufa no Brasil de 1970 a 2021')
```
